```python
DATA_URL = "https://raw.githubusercontent.com/IBM/telco-customer-churn-on-icp4d/master/data/Telco-Customer-Churn.csv"
raw_df = pd.read_csv(DATA_URL)

raw_df["Churn"].value_counts()
```

```python
df = raw_df.copy()

# 1. Target encoding (FIRST, ALWAYS)
df["Churn"] = df["Churn"].map({"Yes": 1, "No": 0}).astype(int)

# 2. Numeric fix
df["TotalCharges"] = pd.to_numeric(df["TotalCharges"], errors="coerce")
df["TotalCharges"] = df["TotalCharges"].fillna(df["TotalCharges"].median())

# 3. Feature engineering
df["charges_per_tenure"] = df["TotalCharges"] / (df["tenure"] + 1)

df = df.sort_values("tenure")
df["rolling_avg_monthly_charges"] = (
    df["MonthlyCharges"].rolling(window=50, min_periods=1).mean()
)

# 4. Binary encoding
binary_cols = ["Partner", "Dependents", "PhoneService", "PaperlessBilling"]
for col in binary_cols:
    df[col] = df[col].map({"Yes": 1, "No": 0})

# 5. One-hot encoding (DO NOT TOUCH TARGET)
df = pd.get_dummies(df, drop_first=True)

# 6. FINAL NaN HANDLING — EXCLUDE TARGET
feature_cols = df.columns.drop("Churn")
df[feature_cols] = df[feature_cols].fillna(0)

# 7. HARD VALIDATION
print("Final target distribution:")
print(df["Churn"].value_counts())
```

```python
TARGET = "Churn"

X = df.drop(columns=[TARGET])
y = df[TARGET].astype(int)

y.value_counts()
```

```python
from sklearn.model_selection import train_test_split

def safe_train_test_split(X, y, test_size=0.25, random_state=42):
    for seed in range(random_state, random_state + 10):
        X_tr, X_te, y_tr, y_te = train_test_split(
            X,
            y,
            test_size=test_size,
            stratify=y,
            random_state=seed
        )
        if y_tr.nunique() > 1:
            return X_tr, X_te, y_tr, y_te
    raise RuntimeError("Unable to create train split with both classes present.")

X_train, X_test, y_train, y_test = safe_train_test_split(X, y)

print("Train class distribution:")
print(y_train.value_counts())
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import roc_auc_score

baseline_model = LogisticRegression(
    max_iter=1000,
    solver="lbfgs"
)

baseline_model.fit(X_train, y_train)

baseline_probs = baseline_model.predict_proba(X_test)[:, 1]
baseline_auc = roc_auc_score(y_test, baseline_probs)

baseline_auc
```

```python
from sklearn.model_selection import StratifiedKFold, cross_val_score

cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

baseline_cv_auc = cross_val_score(
    baseline_model,
    X_train,
    y_train,
    cv=cv,
    scoring="roc_auc"
)

baseline_cv_auc.mean()
```

```python
from xgboost import XGBClassifier
from sklearn.metrics import roc_auc_score

xgb_model = XGBClassifier(
    n_estimators=300,
    max_depth=4,
    learning_rate=0.05,
    subsample=0.8,
    colsample_bytree=0.8,
    eval_metric="logloss",
    random_state=42,
    n_jobs=-1
)

xgb_model.fit(X_train, y_train)

xgb_probs = xgb_model.predict_proba(X_test)[:, 1]
xgb_auc = roc_auc_score(y_test, xgb_probs)

xgb_auc
```

```python
xgb_cv_auc = cross_val_score(
    xgb_model,
    X_train,
    y_train,
    cv=cv,
    scoring="roc_auc"
)

xgb_cv_auc.mean()
```

```python
if xgb_auc >= baseline_auc:
    final_model = xgb_model
    final_probs = xgb_probs
    selected_model = "XGBoost"
else:
    final_model = baseline_model
    final_probs = baseline_probs
    selected_model = "Logistic Regression"

selected_model
```

```python
COST_FALSE_POSITIVE = 50     # retention offered to non-churner
COST_FALSE_NEGATIVE = 500   # churned customer missed
COST_TRUE_POSITIVE = 100    # retention cost
COST_TRUE_NEGATIVE = 0
```

```python
import numpy as np

thresholds = np.linspace(0.05, 0.95, 91)
```

```python
def compute_business_cost(y_true, probs, threshold):
    preds = (probs >= threshold).astype(int)

    tp = np.sum((preds == 1) & (y_true == 1))
    fp = np.sum((preds == 1) & (y_true == 0))
    fn = np.sum((preds == 0) & (y_true == 1))
    tn = np.sum((preds == 0) & (y_true == 0))

    total_cost = (
        tp * COST_TRUE_POSITIVE +
        fp * COST_FALSE_POSITIVE +
        fn * COST_FALSE_NEGATIVE +
        tn * COST_TRUE_NEGATIVE
    )

    return total_cost
```

```python
costs = []

for t in thresholds:
    cost = compute_business_cost(y_test.values, final_probs, t)
    costs.append(cost)

costs = np.array(costs)

optimal_threshold = thresholds[np.argmin(costs)]
optimal_cost = costs.min()

optimal_threshold, optimal_cost
```

```python
baseline_cost = np.sum(y_test == 1) * COST_FALSE_NEGATIVE
baseline_cost
```

```python
business_savings = baseline_cost - optimal_cost
business_savings
```

```python
final_decisions = (final_probs >= optimal_threshold).astype(int)
```

```python
tp = np.sum((final_decisions == 1) & (y_test == 1))
fp = np.sum((final_decisions == 1) & (y_test == 0))
fn = np.sum((final_decisions == 0) & (y_test == 1))
tn = np.sum((final_decisions == 0) & (y_test == 0))

tp, fp, fn, tn
```

```python
decision_df = X_test.copy()
decision_df["actual_churn"] = y_test.values
decision_df["churn_probability"] = final_probs
decision_df["decision"] = final_decisions
```

```python
decision_df["action"] = np.where(
    decision_df["decision"] == 1,
    "RETENTION_ACTION",
    "NO_ACTION"
)

decision_df.head()
```

```python
assert optimal_threshold > 0 and optimal_threshold < 1
assert business_savings >= 0
assert decision_df.shape[0] == X_test.shape[0]

print("STEP 3 COMPLETE — Decision layer executed successfully.")
```

```python
import numpy as np
import pandas as pd
from scipy.stats import ks_2samp
from sklearn.metrics import roc_auc_score
```

```python
reference_df = X_train.copy()
current_df = X_test.copy()

reference_df["prediction"] = final_model.predict_proba(X_train)[:, 1]
current_df["prediction"] = final_probs

reference_df["target"] = y_train.values
current_df["target"] = y_test.values
```

```python
drift_results = []

for col in X_train.columns:
    ref_values = reference_df[col].values
    cur_values = current_df[col].values

    if np.std(ref_values) == 0 or np.std(cur_values) == 0:
        drift_detected = False
    else:
        stat, p_value = ks_2samp(ref_values, cur_values)
        drift_detected = p_value < 0.05

    drift_results.append({
        "feature": col,
        "drift_detected": drift_detected
    })

drift_df = pd.DataFrame(drift_results)
drift_df.head()
```

```python
drifted_features = drift_df["drift_detected"].sum()
total_features = drift_df.shape[0]

drift_share = drifted_features / total_features
dataset_drift = drift_share >= 0.3

dataset_drift, drift_share
```

```python
stat, p_value = ks_2samp(
    reference_df["prediction"],
    current_df["prediction"]
)

prediction_drift = p_value < 0.05
prediction_drift
```

```python
reference_auc = roc_auc_score(
    reference_df["target"],
    reference_df["prediction"]
)

current_auc = roc_auc_score(
    current_df["target"],
    current_df["prediction"]
)

auc_drop = reference_auc - current_auc

reference_auc, current_auc, auc_drop
```

```python
AUC_DROP_THRESHOLD = 0.05
DRIFT_SHARE_THRESHOLD = 0.3

retrain_required = (
    dataset_drift
    or prediction_drift
    or auc_drop >= AUC_DROP_THRESHOLD
)

retrain_required
```

```python
decision_df["dataset_drift_detected"] = dataset_drift
decision_df["prediction_drift_detected"] = prediction_drift
decision_df["retrain_required"] = retrain_required
```

```python
dataset_drift = bool(dataset_drift)
prediction_drift = bool(prediction_drift)
retrain_required = bool(retrain_required)

assert isinstance(dataset_drift, bool)
assert isinstance(prediction_drift, bool)
assert isinstance(retrain_required, bool)

print("STEP 5 COMPLETE — Drift monitoring executed successfully.")
```
